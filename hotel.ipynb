```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df,x="lead_time",y="avg_price_per_room")
plt.title("lead_time vs avg_price_per_room" )
plt.xlabel("lead_time ")
plt.ylabel("avg_price_per_room")

```

```python

data = df.groupby('booking_status', as_index=False)['avg_price_per_room'].mean()

sns.barplot(
    data=data,
    x='booking_status',
    y='avg_price_per_room'
)

plt.xlabel('Booking Status')
plt.ylabel('Average Price per Room')
plt.title('Average Price per Room by Booking Status')

plt.show()
```

```python

data = df['booking_status'].value_counts()

colors = sns.color_palette('Set2')

plt.pie(
    data.values,
    labels=data.index,
    autopct='%1.1f%%',
    colors=colors,
    startangle=90
)

plt.title('Booking Status Distribution')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.lineplot(
    data=df,
    x="arrival_month",
    y="avg_price_per_room",
    marker="o"
)
plt.title("Average Room Price by Arrival Month")
plt.xlabel("Arrival Month")
plt.ylabel("Average Price per Room")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='lead_time', bins=30, kde=True)
plt.title('Distribution of Lead Time')
plt.xlabel('Lead Time')
plt.ylabel('Number of Bookings')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, x='room_type_reserved', hue='booking_status')
plt.title('Room Type vs Booking Status')
plt.xlabel('Room Type')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.violinplot(
    data=df,
    x='booking_status',
    y='no_of_special_requests'
)
plt.title('Special Requests by Booking Status')
plt.xlabel('Booking Status')
plt.ylabel('Special Requests')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.pointplot(
    data=df,
    x='arrival_month',
    y='lead_time',
    hue='booking_status'
)
plt.title('Lead Time by Arrival Month and Booking Status')
plt.xlabel('Arrival Month')
plt.ylabel('Lead Time')
plt.show()
```

```python
plt.figure(figsize=(12,8))

corr = df.select_dtypes(include='number').corr()

sns.heatmap(
    corr,
    cmap='coolwarm',
)

plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.pairplot(
    df,
    vars=[
        'no_of_adults',
        'no_of_children',
        'avg_price_per_room',
        'no_of_special_requests'
    ],
    hue='booking_status'
)

plt.show()
```

```python
sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Booking Status
sns.countplot(
    data=df,
    x='booking_status',
    ax=axes[0, 0]
)
axes[0, 0].set_title('Booking Status Distribution')
axes[0, 0].set_xlabel('Booking Status')
axes[0, 0].set_ylabel('Number of Bookings')


# 2. Market Segment vs Booking Status
sns.countplot(
    data=df,
    x='market_segment_type',
    hue='booking_status',
    ax=axes[0, 1]
)
axes[0, 1].set_title('Market Segment vs Booking Status')
axes[0, 1].set_xlabel('Market Segment')
axes[0, 1].set_ylabel('Number of Bookings')
axes[0, 1].tick_params(axis='x', rotation=45)


# 3. Lead Time Distribution
sns.histplot(
    data=df,
    x='lead_time',
    bins=30,
    kde=True,
    ax=axes[1, 0]
)
axes[1, 0].set_title('Lead Time Distribution')
axes[1, 0].set_xlabel('Lead Time')
axes[1, 0].set_ylabel('Number of Bookings')


# 4. Lead Time vs Average Room Price
sns.scatterplot(
    data=df,
    x='lead_time',
    y='avg_price_per_room',
    hue='booking_status',
    alpha=0.5,
    ax=axes[1, 1]
)
axes[1, 1].set_title('Lead Time vs Average Room Price')
axes[1, 1].set_xlabel('Lead Time')
axes[1, 1].set_ylabel('Average Price per Room')


# Main title
fig.suptitle(
    'Hotel Booking Analysis Dashboard',
    fontsize=20,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid")

# Create 2x2 dashboard
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# ------------------------------------------------
# 1. COUNT PLOT - Booking Status
# ------------------------------------------------
sns.countplot(
    data=df,
    x='booking_status',
    ax=axes[0, 0]
)

axes[0, 0].set_title('Booking Status Distribution', fontsize=14)
axes[0, 0].set_xlabel('Booking Status')
axes[0, 0].set_ylabel('Number of Bookings')


# ------------------------------------------------
# 2. HISTOGRAM - Lead Time
# ------------------------------------------------
sns.histplot(
    data=df,
    x='lead_time',
    bins=30,
    kde=True,
    ax=axes[0, 1]
)

axes[0, 1].set_title('Lead Time Distribution', fontsize=14)
axes[0, 1].set_xlabel('Lead Time')
axes[0, 1].set_ylabel('Number of Bookings')


# ------------------------------------------------
# 3. SCATTER PLOT - Lead Time vs Price
# ------------------------------------------------
sns.scatterplot(
    data=df,
    x='lead_time',
    y='avg_price_per_room',
    hue='booking_status',
    alpha=0.6,
    ax=axes[1, 0]
)

axes[1, 0].set_title(
    'Lead Time vs Average Room Price',
    fontsize=14
)
axes[1, 0].set_xlabel('Lead Time')
axes[1, 0].set_ylabel('Average Price per Room')


# ------------------------------------------------
# 4. HEATMAP - Correlation
# ------------------------------------------------
numeric_df = df.select_dtypes(include='number')

corr = numeric_df.corr()

sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    ax=axes[1, 1]
)

axes[1, 1].set_title(
    'Correlation Heatmap',
    fontsize=14
)


# ------------------------------------------------
# DASHBOARD TITLE
# ------------------------------------------------
fig.suptitle(
    'Hotel Booking Analysis Dashboard',
    fontsize=22,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid")

# Create 3x2 dashboard
fig, axes = plt.subplots(3, 2, figsize=(18, 18))

# ------------------------------------------------
# 1. COUNT PLOT
# Booking Status
# ------------------------------------------------
sns.countplot(
    data=df,
    x='booking_status',
    ax=axes[0, 0]
)

axes[0, 0].set_title('Booking Status Distribution', fontsize=14)
axes[0, 0].set_xlabel('Booking Status')
axes[0, 0].set_ylabel('Number of Bookings')


# ------------------------------------------------
# 2. HISTOGRAM
# Lead Time
# ------------------------------------------------
sns.histplot(
    data=df,
    x='lead_time',
    bins=30,
    kde=True,
    ax=axes[0, 1]
)

axes[0, 1].set_title('Lead Time Distribution', fontsize=14)
axes[0, 1].set_xlabel('Lead Time')
axes[0, 1].set_ylabel('Number of Bookings')


# ------------------------------------------------
# 3. BOXPLOT
# Average Price by Booking Status
# ------------------------------------------------
sns.boxplot(
    data=df,
    x='booking_status',
    y='avg_price_per_room',
    ax=axes[1, 0]
)

axes[1, 0].set_title(
    'Average Room Price by Booking Status',
    fontsize=14
)
axes[1, 0].set_xlabel('Booking Status')
axes[1, 0].set_ylabel('Average Price per Room')


# ------------------------------------------------
# 4. VIOLIN PLOT
# Special Requests by Booking Status
# ------------------------------------------------
sns.violinplot(
    data=df,
    x='booking_status',
    y='no_of_special_requests',
    ax=axes[1, 1]
)

axes[1, 1].set_title(
    'Special Requests by Booking Status',
    fontsize=14
)
axes[1, 1].set_xlabel('Booking Status')
axes[1, 1].set_ylabel('Number of Special Requests')


# ------------------------------------------------
# 5. SCATTER PLOT
# Lead Time vs Average Price
# ------------------------------------------------
sns.scatterplot(
    data=df,
    x='lead_time',
    y='avg_price_per_room',
    hue='booking_status',
    alpha=0.5,
    ax=axes[2, 0]
)

axes[2, 0].set_title(
    'Lead Time vs Average Room Price',
    fontsize=14
)
axes[2, 0].set_xlabel('Lead Time')
axes[2, 0].set_ylabel('Average Price per Room')


# ------------------------------------------------
# 6. HEATMAP
# Numerical Correlation
# ------------------------------------------------
numeric_df = df.select_dtypes(include='number')

corr = numeric_df.corr()

sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    ax=axes[2, 1]
)

axes[2, 1].set_title(
    'Correlation Heatmap',
    fontsize=14
)


# ------------------------------------------------
# DASHBOARD TITLE
# ------------------------------------------------
fig.suptitle(
    'Hotel Booking Analysis Dashboard',
    fontsize=24,
    fontweight='bold',
    y=1.02
)

plt.tight_layout()
plt.show()
```
